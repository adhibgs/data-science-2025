Langkah 1: Muat dan Eksplorasi Data


```python
import pandas as pd
df = pd.read_csv("/content/sample_data/Telco-Customer-Churn.csv")
print(df.shape)
print(df["Churn"].value_counts(normalize=True))
```

Langkah 2: Preprocessing


```python
from sklearn.model_selection import train_test_split

# Mengubah nilai 'Yes'/'No' pada kolom 'Churn' menjadi 1/0
df['Churn'] = df['Churn'].apply(lambda x: 1 if x == 'Yes' else 0)

# Menghilangkan kolom 'customerID' karena tidak relevan sebagai fitur
df_processed = df.drop('customerID', axis=1)

# Melakukan encoding fitur kategorikal menggunakan pd.get_dummies
df_processed = pd.get_dummies(df_processed, drop_first=True)

# Memisahkan X (fitur) dan y (target = Churn)
X = df_processed.drop('Churn', axis=1)
y = df_processed['Churn']

X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42)
```

Langkah 3: Latih Model


```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=300, class_weight="balanced", random_state=42)
rf.fit(X_tr, y_tr)
```

Langkah 4: Evaluasi


```python
from sklearn.metrics import classification_report, roc_auc_score
y_pred_rf = rf.predict(X_te)
y_proba_rf = rf.predict_proba(X_te)[:, 1]
print("=== Classification Report ===")
print(classification_report(y_te, y_pred_rf))
print(f"ROC-AUC Score: {roc_auc_score(y_te, y_proba_rf):.4f}")
# TODO: hitung prediksi dan probabilitas
# TODO: tampilkan classification_report dan ROC-AUC
```

Langkah 5: Prediksi Probabilitas dan Simpulkan


```python
sample_proba = pd.DataFrame({
    'Status Churn Aktual': y_te[:5].values,
    'Prediksi Churn': y_pred_rf[:5],
    'Probabilitas Churn (%)': (y_proba_rf[:5] * 100).round(2)
})
print("\n=== Contoh Prediksi Probabilitas Churn ===")
print(sample_proba)
# TODO: hitung probabilitas churn (predict_proba)
# TODO: tulis kesimpulan singkat (3-4 kalimat)
```
